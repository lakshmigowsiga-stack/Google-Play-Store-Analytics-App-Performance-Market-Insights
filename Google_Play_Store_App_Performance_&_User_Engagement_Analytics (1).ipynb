# Stage 2 – Data Cleaning and Pre-processing

**Clean Rating**

```python
# Convert Rating to numeric
df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")
```

```python
# Check invalid ratings
invalid_ratings = df[(df["Rating"] < 1) | (df["Rating"] > 5)]
```

```python
print("Invalid rating rows:", len(invalid_ratings))
display(invalid_ratings)
```

```python
# Remove invalid rating rows

df = df[
    df["Rating"].between(1, 5) |
    df["Rating"].isna()].copy()
print("Dataset shape after removing invalid ratings:", df.shape)
```

# Remove Duplicate Rows

```python
# Remove duplicate rows
print("Duplicate rows before removal:", df.duplicated().sum())
df = df.drop_duplicates().reset_index(drop=True)
print("Duplicate rows after removal:", df.duplicated().sum())
print("Dataset shape:", df.shape)
```

# Clean App and Category

```python
# Clean text columns
df["App"] = df["App"].astype(str).str.strip()
df["Category"] = df["Category"].astype(str).str.strip()
print("App and Category columns cleaned.")
```

# Handle Missing Ratings

```python
# Fill missing ratings using the median
rating_median = df["Rating"].median()
df["Rating"] = df["Rating"].fillna(rating_median)
print("Missing ratings after cleaning:", df["Rating"].isnull().sum())
print("Rating median used:", rating_median)
```

# Clean Reviews

```python
# Convert Reviews to numeric

df["Reviews"] = pd.to_numeric(
    df["Reviews"],
    errors="coerce")

# Replace missing values with 0
df["Reviews"] = df["Reviews"].fillna(0).astype("int64")

print("Reviews column cleaned.")
print(df["Reviews"].dtype)
```

# Clean Size

```python
# Function to convert app size into MB

def convert_size(size):

    size = str(size).strip()

    if size == "Varies with device":
        return np.nan

    if size.endswith("M"):
        return float(size.replace("M", ""))

    if size.endswith("k"):
        return float(size.replace("k", "")) / 1024

    return np.nan


# Apply conversion
df["Size"] = df["Size"].apply(convert_size)

# Fill missing sizes with median
df["Size"] = df["Size"].fillna(df["Size"].median())

print("Size column converted to MB.")
print(df["Size"].describe())
```

# Clean Installs

```python
# Clean Installs column

df["Installs"] = (
    df["Installs"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False)
    .str.strip())

# Convert to numeric
df["Installs"] = pd.to_numeric(
    df["Installs"],
    errors="coerce")

# Fill missing values
df["Installs"] = df["Installs"].fillna(0).astype("int64")

print("Installs column cleaned.")
print(df["Installs"].describe())
```

```python
# Check installation values

print("Minimum installs:", df["Installs"].min())
print("Maximum installs:", df["Installs"].max())

display(df[["App", "Installs"]].head(10))
```

# Clean Type

```python
# Handle missing Type values

df["Type"] = df["Type"].fillna(
    df["Type"].mode()[0])

df["Type"] = df["Type"].astype(str).str.strip()

print("App Type values:")
print(df["Type"].value_counts())
```

# Clean Price

```python
# Clean Price column

df["Price"] = (
    df["Price"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.strip())

# Convert to numeric
df["Price"] = pd.to_numeric(
    df["Price"],
    errors="coerce")

# Fill missing prices
df["Price"] = df["Price"].fillna(0).astype(float)

print("Price column cleaned.")
print(df["Price"].describe())
```

```python
# Check free and paid apps

print("Free and Paid Apps:")
print(df["Type"].value_counts())

print("\nNumber of paid apps:")
print((df["Price"] > 0).sum())

print("\nHighest app price:")
print(df["Price"].max())
```

# Clean Content Rating

```python
# Clean Content Rating

df["Content Rating"] = df["Content Rating"].fillna(
    df["Content Rating"].mode()[0])

df["Content Rating"] = (
    df["Content Rating"]
    .astype(str)
    .str.strip())

print(df["Content Rating"].value_counts())
```

# Clean Genres

```python
# Clean Genres

df["Genres"] = (
    df["Genres"]
    .astype(str)
    .str.strip())

print("Number of unique genres:", df["Genres"].nunique())
```

# Clean Last Updated

```python
# Convert Last Updated to datetime

df["Last Updated"] = pd.to_datetime(
    df["Last Updated"],
    errors="coerce")

print(df["Last Updated"].dtype)
display(df[["App", "Last Updated"]].head())
```

# Clean Version Columns

```python
# Clean Current Version

df["Current Ver"] = (
    df["Current Ver"]
    .fillna("Unknown")
    .astype(str)
    .str.strip())

# Clean Android Version

df["Android Ver"] = (
    df["Android Ver"]
    .fillna("Unknown")
    .astype(str)
    .str.strip())

print("Version columns cleaned.")
```

```python
# Check missing values after cleaning

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# Check duplicate rows

print("Duplicate rows:", df.duplicated().sum())
```

# Feature Engineering

**Create Updated Year and Month**

```python
# Extract year and month from Last Updated

df["Updated_Year"] = df["Last Updated"].dt.year
df["Updated_Month"] = df["Last Updated"].dt.month

print("Updated year and month columns created.")

display(df[["App", "Last Updated", "Updated_Year", "Updated_Month"]].head())
```

# Create Review-to-Install Ratio

```python
# Calculate Review-to-Install Ratio

df["Review_Install_Ratio"] = np.where(
    df["Installs"] > 0,
    df["Reviews"] / df["Installs"],
    np.nan)

print("Review-to-Install Ratio created.")

display(df[["App", "Reviews", "Installs", "Review_Install_Ratio"]].head(10))
```

# Create Popularity Category

```python
# Create popularity categories based on installs

df["Popularity_category"] = np.where(
    df["Installs"] < 10000,
    "Less popular",
    np.where(
        df["Installs"] < 500000,
        "Medium popular",
        "Most popular"))

print(df["Popularity_category"].value_counts())
```

## 5.correlation heatmap

```python
numeric_columns = [
    "Rating",
    "Reviews",
    "Installs",
    "Price",
    "Size",
    "Review_Install_Ratio"]
corr = df[numeric_columns].corr()
plt.figure(figsize=(9, 7))
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    linewidths=0.5)
plt.title("Correlation Heatmap of Numerical Features",fontsize=14)
plt.tight_layout()
plt.show()
```

# Interpretation for above chart:
1.The correlation heatmap shows the relationships between the numerical features in the Google Play Store dataset. The values range from -1 to +1, where values closer to +1 indicate a positive relationship, values closer to -1 indicate a negative relationship, and values close to 0 indicate a weak linear relationship.

2.The heatmap helps identify which numerical variables are strongly or weakly related to each other.

3.corr() → calculates the correlation between numerical variables.
  
  annot=True → displays the correlation values inside the boxes.
  
  cmap="coolwarm" → gives different shades based on correlation strength.
  
  fmt=".2f" → displays values with 2 decimal places.

4.The correlation heatmap provides an overall view of the relationships between numerical variables and helps identify important relationships that can be investigated further using scatter plots and other visualizations.
